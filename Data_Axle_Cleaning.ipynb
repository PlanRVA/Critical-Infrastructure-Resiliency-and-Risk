```python
health = gpd.read_file(r'')

# Print initial count
print("Initial row count:", len(health))

# Ensure NAICS is a string and create NAICS6 (first 6 digits)
health["NAICS6"] = (
    health["NAICS"]
    .astype(str)
    .str.strip()
    .str[:6]
)


Health_NAICS = ["621111", "611699", "611710", "621112", "621210", 
                "621310", "621320", "621330", "621340", "621391", "621399",
                "621420", "621491", "621492", "621493", "621498", "621511", "621512",
                "621610", "621910", "621999", "622110", "622210", "622310", "623110", "541940"]
                    
health2 = health[health["NAICS6"].isin(Health_NAICS)]

# Print final count
print("Final row count:", len(health2))

# Save result to point data cleaned
health2.to_file("")
```

```python
health2 = gpd.read_file(r'')

#Things we want in the HexBins are the Sector and a Sub-Sector Category
#Want to make a column "SECTOR" that is "Healthcare and Public Health"
#Want to make a Sub-Sector Column "S_SECTOR" with mapped sub-sectors based on NAICS
#What might fall under "Health Education" ? NAICs for CPR, and First Aid education or training
#Substance Abuse falls under Outpatient Mental Health.

health2["Sector"] = "Healthcare and Public Health"

subsector_categories  = {"621111": "Physicians", "611699": "Health Education", "611710": "Health Education", "621112": "Mental Health",
                         "621210": "Dentists", "621310": "Chiropractors", "621320": "Optometrists", "621330": "Mental Health",
                         "621340": "Physical Therapy", "621391": "Podiatrists", "621399": "Health Practitioners", "621420": "Outpatient Mental Health",
                         "621491": "HMO Medical Center", "621492": "Dialysis Center", "621493": "Emergency Center", "621498": "Outpatient Care Centers",
                         "621511": "Medical Laboratories", "621512": "Dianostic Imaging", "621610": "Home Health Care", "621910": "Ambulance Services",
                         "621999": "Ambulance Services", "622110": "Hospitals", "622210": "Psychiatric Hospitals", "622310": "Specialty Hospitals",
                         "623110": "Nursing Care Facilities", "541940": "Veterinary Services"}

health2["Sub-Sector"] = health2["NAICS6"].map(subsector_categories)
# Save result to point data cleaned
health2.to_file("")
```
