### Initial Cleaning of Polygon Data
#### Assigning Sector and Sub-Sector Categories

```python
import geopandas as gpd
import pandas as pd
from shapely.geometry import Point
import numpy as np
```

### Water Sector

```python
#input data of water reservoirs
#in a folder for raw input data
reservoirs = gpd.read_file(r'')
water = gpd.read_file(r'')

#Assign Sectors and Sub-Sectors
reservoirs["Sector"] = "Water and Wastewater"
reservoirs["Sub-Sector"] = "Water Reservoirs"


#Reduce Columns
reservoirs = reservoirs[["Sector", "Sub-Sector", "geometry"]]

# Make sure CRS matches -
# here we are going to make it match the water line data we just did
target_crs = water.crs

datasets = [
    reservoirs
]

datasets = [gdf.to_crs(target_crs) for gdf in datasets]

#output cleaned polygon data - 
reservoirs.to_file(r'')
```
