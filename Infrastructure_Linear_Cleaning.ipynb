### Transportation Sector

```python
#read all line data - these were in various folders. Can edit to read all within a folder:
strahnet = gpd.read_file(r'')
evac = gpd.read_file(r'')
runways = gpd.read_file(r'')
rail = gpd.read_file(r'')
hazards = gpd.read_file(r'')
foot = gpd.read_file(r'')
culverts = gpd.read_file(r'')
bridges = gpd.read_file(r'')
roads = gpd.read_file(r'')
bus = gpd.read_file(r'')


#Assign Sectors and Sub-Sectors
strahnet["Sector"] = "Transportation"
strahnet["Sub-Sector"] = "Strategic Highway Network"
evac["Sector"] = "Transportation"
evac["Sub-Sector"] = "Hurricane Evacuation Route"
runways["Sector"] = "Transportation"
runways["Sub-Sector"] = "Aviation Runways"
rail["Sector"] = "Transportation"
rail["Sub-Sector"] = "Rail Lines"
hazards["Sector"] = "Transportation"
hazards["Sub-Sector"] = "Hazardous Materials Routes"
foot["Sector"] = "Transportation"
foot["Sub-Sector"] = "Footbridges"
culverts["Sector"] = "Transportation"
culverts["Sub-Sector"] = "Culverts"
bridges["Sector"] = "Transportation"
bridges["Sub-Sector"] = "Bridges"
roads["Sector"] = "Transportation"
roads["Sub-Sector"] = "Road Network"
bus["Sector"] = "Transportation"
bus["Sub-Sector"] = "Bus Network"

#Reduce Columns
strahnet = strahnet[["Sector", "Sub-Sector", "geometry"]]
evac = evac[["Sector", "Sub-Sector", "geometry"]]
runways = runways[["Sector", "Sub-Sector", "geometry"]]
rail = rail[["Sector", "Sub-Sector", "geometry"]]
hazards = hazards[["Sector", "Sub-Sector", "geometry"]]
foot = foot[["Sector", "Sub-Sector", "geometry"]]
culverts = culverts[["Sector", "Sub-Sector", "geometry"]]
bridges = bridges[["Sector", "Sub-Sector", "geometry"]]
roads = roads[["Sector", "Sub-Sector", "geometry"]]
bus = bus[["Sector", "Sub-Sector", "geometry"]]

# Make sure CRS matches
target_crs = roads.crs

datasets = [
    strahnet,
    evac,
    runways,
    rail,
    hazards,
    foot,
    culverts,
    bridges,
    roads,
    bus
]

datasets = [gdf.to_crs(target_crs) for gdf in datasets]

# Merge ALL
transportation1 = gpd.GeoDataFrame(
    pd.concat(datasets, ignore_index=True),
    crs=target_crs
)

#saving where all linear cleaned data is going:
transportation1.to_file(r'')
```
