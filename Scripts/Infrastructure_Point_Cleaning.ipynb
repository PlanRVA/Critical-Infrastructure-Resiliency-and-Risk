### Government Facilities

```python
cova = gpd.read_file(r'')
publicschools = gpd.read_file(r'')
highered = gpd.read_file(r'')
libraries = gpd.read_file(r'')
private = gpd.read_file(r'')
usps1 = gpd.read_file(r'')
usps2 = gpd.read_file(r'')
bus = gpd.read_file(r'')
fbi = gpd.read_file(r'')
waste = gpd.read_file(r'')
court = gpd.read_file(r'')

#assign sectors and sub sectors as needed
cova["Sector"] = "Government Facilities"
cova["Sub-Sector"] = "State Agency Buildings"
publicschools["Sector"] = "Government Facilities"
publicschools["Sub-Sector"] = "Public Schools"
highered["Sector"] = "Government Facilities"
highered["Sub-Sector"] = "Higher Education"
libraries["Sector"] = "Government Facilities"
libraries["Sub-Sector"] = "Public Libraries"
private["Sector"] = "Government Facilities"
private["Sub-Sector"] = "Private Schools"
usps1["Sector"] = "Government Facilities"
usps1["Sub-Sector"] = "USPS Locations"
usps2["Sector"] = "Government Facilities"
usps2["Sub-Sector"] = "USPS Plants"
bus["Sector"] = "Government Facilities"
bus["Sub-Sector"] = "School Bus Storage"
fbi["Sector"] = "Government Facilities"
fbi["Sub-Sector"] = "Regional FBI Building"
waste["Sector"] = "Government Facilities"
waste["Sub-Sector"] = "Waste and Recycling Facilities"
court["Sector"] = "Government Facilities"
court["Sub-Sector"] = "Courthouses"

#read only data we want
cova = cova[["Sector", "Sub-Sector", "geometry"]]
publicschools = publicschools[["Sector", "Sub-Sector", "geometry"]]
highered = highered[["Sector", "Sub-Sector", "geometry"]]
libraries = libraries[["Sector", "Sub-Sector", "geometry"]]
private = private[["Sector", "Sub-Sector", "geometry"]]
usps1 = usps1[["Sector", "Sub-Sector", "geometry"]]
usps2 = usps2[["Sector", "Sub-Sector", "geometry"]]
bus = bus[["Sector", "Sub-Sector", "geometry"]]
fbi = fbi[["Sector", "Sub-Sector", "geometry"]]
waste = waste[["Sector", "Sub-Sector", "geometry"]]
court = court[["Sector", "Sub-Sector", "geometry"]]

# Make sure CRS matches
target_crs = cova.crs

datasets = [
    cova,
    publicschools,
    highered,
    libraries,
    private,
    usps1,
    usps2,
    bus,
    fbi,
    waste,
    court
]

# Reproject safely
datasets = [gdf.to_crs(target_crs) for gdf in datasets]

# Merge
gov = gpd.GeoDataFrame(
    pd.concat(datasets, ignore_index=True),
    crs=target_crs
)

# Clean
gov = gov[gov.geometry.notnull()]
gov = gov[~gov.geometry.is_empty]

def finite_geom(geom):
    if geom is None or geom.is_empty:
        return None
    try:
        coords = np.array(geom.coords)
        if not np.isfinite(coords).all():
            return None
    except Exception:
        return None
    return geom
# Remove non-finite coordinates
gov["geometry"] = gov.geometry.apply(finite_geom)
gov = gov.dropna(subset=["geometry"])

# Save
gov.to_file(
    r""
)
```
