## Final Data to Parquet
##### Taking the final data and making a parquet file for the dashboard

```python
import requests
import pandas as pd
import geopandas as gpd
from shapely.geometry import shape
import h3
```

```python

# Read directly from shapefile
gdf = gpd.read_file(r"")

print(f"Loaded {len(gdf)} rows")
print(gdf.columns.tolist())
print(gdf.crs)

# Reproject to WGS84 if needed
if gdf.crs.to_epsg() != 4326:
    gdf = gdf.to_crs("EPSG:4326")
    print("Reprojected to EPSG:4326")

# Save as parquet
output = r".parquet"
gdf.to_parquet(output)
print(f"Saved to {output}")
```
